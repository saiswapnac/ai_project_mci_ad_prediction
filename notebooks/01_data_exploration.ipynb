# 01 — ADNI clinical cohort construction

This notebook runs the local cohort builder for the separate ADNI exports (`DXSUM`, `PTDEMOG`, `MMSE`, and `ADAS`). It does not require MRI files yet. Keep participant-level data and outputs on your approved local storage; do not upload them to public-facing AI tools.

The script uses the supplied dictionary mappings: `DXMCI`/`DXAD`/`DXNORM` are checkbox fields (`1=Yes`), `PTGENDER` is `1=Male`/`2=Female`, MMSE is `MMSCORE`, and the preferred ADAS score is `TOTAL13`. Review output definitions and aggregate counts before modeling.


```python
from pathlib import Path
import subprocess, sys

script_name = Path('scripts/build_cohort_from_tables.py')
candidates = [Path.cwd(), *Path.cwd().parents]
project_root = next((p for p in candidates if (p / script_name).exists()), None)
if project_root is None:
    raise FileNotFoundError('Could not find scripts/build_cohort_from_tables.py. Open the mci_ad_prediction folder in VS Code and rerun this cell.')
result = subprocess.run([sys.executable, str(project_root / script_name)], cwd=project_root, check=False)
if result.returncode != 0:
    raise RuntimeError(f'Cohort builder exited with code {result.returncode}; inspect the terminal output.')
```
